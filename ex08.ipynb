```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# =========================
# 1. DATA
# =========================
texts = [
    "I love this movie",
    "This film is amazing",
    "Amazing acting",
    "I really like this film",
    "I hate this movie",
    "This movie is bad",
    "Worst film ever",
    "I dislike this movie"
]

labels = np.array([1,1,1,1,0,0,0,0])

# =========================
# 2. TOKENIZATION
# =========================
tok = Tokenizer()

tok.fit_on_texts(texts)

# Convert text to Bag of Words
X = tok.texts_to_matrix(texts, mode='count')

print("BoW Matrix:")
print(X)

# =========================
# 3. MODEL
# =========================
model = Sequential([
    Dense(8, activation='relu', input_shape=(X.shape[1],)),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# =========================
# 4. TRAIN
# =========================
model.fit(X, labels, epochs=100)

# =========================
# 5. EVALUATE
# =========================
loss, acc = model.evaluate(X, labels)

print("Accuracy:", acc)

# =========================
# 6. TEST
# =========================
test = [
    "this movie is amazing",
    "this is bad"
]

X_test = tok.texts_to_matrix(test, mode='count')

pred = model.predict(X_test)

print("Predictions:")
print(pred)
```
