```python
import numpy as np
```

```python
states=["rainy","sunny"]
states_idx={s: i for i,s in enumerate(states)}
print(states_idx)
```

```python
start_prob=np.array([0.6,0.4])

trans_prob=np.array([
    [0.7,0.3],
    [0.4,0.6]
    ])
emit_prob=np.array([
    [0.1,0.4,0.5],
    [0.6,0.3,0.1]
    ])
```

```python
obs=["walk","sleep","clean"]
obs_idx={"walk":0,"sleep":1,"clean":2}

obs_seq=[obs_idx[o] for o in obs_idx]
print(obs_seq)
T=len(obs_seq)
N=len(states)
```

```python
alpha=np.zeros((T,N))
alpha[0]=start_prob*emit_prob[:,obs_seq[0]]
print("A1 =",alpha[0])
```

```python
for t in range(1,T):
    for j in range(N):
        alpha[t,j]=np.sum(alpha[t-1]*trans_prob[:,j])*emit_prob[j,obs_seq[t]]
    print(f"A{t+1} =",alpha[t])
```

```python
p=np.sum(alpha[-1])
print("forward prob =",p)
```

```python
beta=np.zeros((T,N))
beta[T-1]=1
print("bT =",beta[T-1])
```

```python
for t in range(T-2,-1,-1):
    for j in range(N):
        beta[t,j]=np.sum(trans_prob[j,:]*emit_prob[:,obs_seq[t+1]]*beta[t+1])
    print(f"b{t+1} =",beta[t])
```

```python
p=np.sum(start_prob*emit_prob[:,obs_seq[0]]*beta[0])
print("backward prob=",p)
```
