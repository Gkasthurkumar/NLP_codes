```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.util import ngrams
from collections import Counter
```

```python
content="""
Social media platforms like instagram twitter and tiktok dominate daily screen time for millions of people around the world.
Short form video content has completely changed how we consume entertainment these days.

Remote work became the new normal after the global pandemic forced companies to adapt quickly.
Video conferencing tools such as zoom microsoft teams and google meet are now essential parts of professional life.
Many employees prefer hybrid work models that combine office days with work from home flexibility.

Online shopping continues to grow rapidly with amazon flipkart and other e commerce giants leading the market.
Fast delivery services same day delivery and easy returns have raised customer expectations significantly.

Mobile gaming exploded in popularity thanks to free to play titles and in app purchases.
Battle royale games like pubg fortnite and call of duty mobile attract huge player communities.
Esports tournaments now offer prize pools worth millions of dollars every year.

Streaming services including netflix disney plus amazon prime video and youtube premium compete fiercely for subscribers.
Binge watching entire seasons over one weekend has become a common weekend activity.

Cryptocurrency trading and nfts gained massive attention during the bull market of twenty twenty one.
Blockchain technology promises to revolutionize finance supply chains and digital ownership rights.

Electric vehicles from tesla byd and other manufacturers are reshaping the automotive industry.
Charging stations battery range and fast charging technology remain key discussion points among buyers.

Mental health awareness has increased dramatically in recent years especially among younger generations.
Meditation apps therapy platforms and online support communities provide accessible help for many people.

Artificial intelligence assistants like siri alexa and google assistant handle everyday tasks with voice commands.
Generative ai tools can now create realistic images write essays and even compose music in seconds.

Sustainable living practices such as zero waste minimalism and plant based diets are gaining more followers every month.
Climate change discussions dominate news headlines and influence government policies worldwide.

"""
```

```python
tokens=nltk.word_tokenize(content)
N=len(tokens)
```

```python
unigrams=list(ngrams(tokens,1))
bigrams=list(ngrams(tokens,2))

unigrams_freq=Counter(unigrams)
bigrams_freq=Counter(bigrams)
#w1 is the tuple in freq
#dont forget to give [(w1,)] in next cell
```

```python
for bigram,a in bigrams_freq.items():
    w1,w2=bigram
    b=unigrams_freq[(w1,)]-a
    c=unigrams_freq[(w2,)]-a
    d=N-(a+b+c)

    row1=a+b
    row2=c+d
    col1=a+c
    col2=b+d

    ex_a=(row1*col1)/N
    ex_b=(row1*col2)/N
    ex_c=(row2*col1)/N
    ex_d=(row2*col2)/N
    if ex_a ==0 or ex_b ==0 or ex_c ==0 or ex_d ==0:
        continue
    chi_sq=(ex_a-a)**2/ex_a +\
            (ex_b-b)**2/ex_b+\
            (ex_c-c)**2/ex_c+\
            (ex_d-d)**2/ex_d
    
    if chi_sq>3.84:
        print(w1,w2)
```

```python
#t test 
import math
for bigram ,o in bigrams_freq.items():
    w1,w2=bigram
    e=(unigrams_freq[(w1,)]*unigrams_freq[(w2,)])/N
    if o==0 or e==0:
        continue
    t_value=(o-e)/math.sqrt(o)
    if(t_value>1):
        print(w1,w2)
```
