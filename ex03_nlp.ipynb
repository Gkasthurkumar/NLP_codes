```python
from collections import Counter
import math
import pandas as pd
```

```python
df=pd.read_csv("dataset3.csv")
df.head()
```

```python
texts=[]
labels=[]
for i in range(len(df)):
    sentence=str(df['Text'][i].lower().split())
    label=df['Label'][i]
    texts.append(sentence)
    labels.append(label)
```

```python
vocab=set()
for sentence in texts:
    for word in sentence:
        vocab.add(word)
vocab=list(vocab)
print("vocab size =",len(vocab))
```

```python
labels_count=Counter(labels)
print(labels_count)
```

```python
word_counts={}
for label in set(labels):
    word_counts[label]=Counter()
```

```python
for sentence,label in zip(texts,labels):
    for word in sentence:
        word_counts[label][word]+=1
```

```python
def predict(sentence):
    words=sentence.lower().split()
    scores={}
    vocab_size=len(vocab)
    for label in labels_count:
        score= labels_count[label]/len(labels)
        total_words=sum(word_counts[label].values())
        for word in words:
            word_prob=(word_counts[label][word]+1)/(total_words+vocab_size)
            score*=word_prob
        scores[label]=score
    predicted=max(scores,key=scores.get)
    return predicted
```

```python
test = "government election parliament minister"

result = predict(test)

print("Test Sentence:")
print(test)

print("\nPredicted Label:")
print(result)
```
