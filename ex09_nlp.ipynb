```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
df=pd.read_csv("dataset07.csv")
df.head()
```

```python
df['Category']=df['Category'].map({
    'ham' :0,
    'spam':1
    })
```

```python
message=[]
for text in df['Message']:
    text=text.lower()
    words=text.split()
    message.append(words)
labels=list(df["Category"])
```

```python
vocab=sorted(set(
    word
    for sentence in message
    for word in sentence
))

print("Vocab size =",len(vocab))
```

```python
#bag of words
#bow function
def bow_vector(sentence,vocab):
    vector=[]
    word_count=Counter(sentence)
    for word in vocab:
        vector.append(word_count[word])
    return vector
```

```python
bow_vectors=[]
for sentence in message[:5]:
    vec=bow_vector(sentence,vocab)
    bow_vectors.append(vec)
    print("Sentence")
    print(sentence)
    print(sum(vec))
    print("BoW vector")
    print(vec[:20])
```

```python
#TF-IDF

N=len(message)

dfeq={}
for word in vocab:
    count=0
    for sentence in message:
        if word in sentence:
            count+=1
    dfeq[word]=count
```

```python
idf={}
for word in vocab:
    idf[word]=np.log(N/dfeq[word])
```

```python
def tfidf_vector(sentence):
    word_count=Counter(sentence)
    sentence_len=len(sentence)
    vector=[]
    for word in vocab:
        tf=word_count[word]/sentence_len
        tfid=tf*idf[word]
        vector.append(tfid)
    return vector
```

```python
print("TF-IDF vectors")
for sentence in message[:3]:
    vec=tfidf_vector(sentence)
    print(sentence)
    print(vec[:20])
```

```python
embedding_size=5
word_embedding={}
for word in vocab:
    word_embedding[word]=np.random.rand(embedding_size)
```

```python
def sentence_vector(sentence):
    vectors=[]
    for word in sentence:
        vectors.append(word_embedding[word])
    return np.mean(vectors,axis=0)
```

```python
print("\nWORD2VEC REPRESENTATION\n")

for sentence in message[:3]:
    vec = sentence_vector(sentence)
    print(sentence)
    print(vec)
```

```python
spam_words=[]
ham_words=[]
for sentence,label in zip(message,labels):
    if label==1:
        spam_words.extend(sentence)
    else:
        ham_words.extend(sentence)
```

```python
spam_count = Counter(spam_words)
ham_count = Counter(ham_words)
```

```python
def predict(sentence):

    sentence = sentence.lower().split()

    spam_score = 1
    ham_score = 1

    for word in sentence:

        spam_score *= (spam_count[word] + 1)

        ham_score *= (ham_count[word] + 1)

    if spam_score > ham_score:
        return "SPAM"

    return "HAM"
```

```python
test = "Free prize you have to claim now"

result = predict(test)

print("\nTest Message =", test)

print("Prediction =", result)
```
