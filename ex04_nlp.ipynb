```python
import nltk
from collections import Counter
import math

# =========================
# 1. CORPUS + TOKENIZATION
# =========================
text = """
John saw the man with a telescope
He ate lunch with a spoon
She saw a dog in the park
The boy saw the girl with a camera
"""

tokens = nltk.word_tokenize(text.lower())

print("Tokens:\n", tokens)

# =========================
# 2. UNIGRAM & BIGRAM COUNTS
# =========================
unigram = Counter(tokens)
bigram = Counter(nltk.bigrams(tokens))

print("\nUnigram Counts:\n", unigram)
print("\nBigram Counts:\n", bigram)

# =========================
# 3. USER INPUT
# =========================
n = input("\nEnter noun: ").lower()
v = input("Enter verb: ").lower()
p = input("Enter preposition: ").lower()

print("\nInput → Noun:", n, "Verb:", v, "Preposition:", p)

# =========================
# 4. GET COUNTS
# =========================
Cn = unigram[n]
Cv = unigram[v]

Cpn = bigram[(p, n)]
Cpv = bigram[(p, v)]

print("\nCounts:")
print("C(n):", Cn)
print("C(v):", Cv)
print("C(p,n):", Cpn)
print("C(p,v):", Cpv)

# =========================
# 5. PROBABILITIES (with smoothing)
# =========================
epsilon = 1e-6

Pv = (Cpv + epsilon) / (Cv + epsilon)
Pn = (Cpn + epsilon) / (Cn + epsilon)

print("\nProbabilities:")
print("P(p|v):", Pv)
print("P(p|n):", Pn)

# =========================
# 6. DECISION
# =========================
if Pv > Pn:
    print("\nAttachment: VERB")
else:
    print("\nAttachment: NOUN")
```
