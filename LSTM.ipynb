```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# -------- Data --------
text = "i love  machine learning and deep in the ass "

tok = Tokenizer()
tok.fit_on_texts([text])
seq = tok.texts_to_sequences([text])[0]

X, y = [], []
for i in range(1, len(seq)):
    X.append(seq[:i])
    y.append(seq[i])

X = pad_sequences(X)

# ✅ Convert to dataset (FIX)
dataset = tf.data.Dataset.from_tensor_slices((X, y)).batch(2)

# -------- Model --------
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X.shape[1],)),   # important
    tf.keras.layers.Embedding(len(tok.word_index)+1, 8),
    tf.keras.layers.LSTM(10),
    tf.keras.layers.Dense(len(tok.word_index)+1, activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')
model.fit(dataset, epochs=100, verbose=1)

# -------- Generate --------
seed = "i love deep"

for _ in range(5):
    seq = pad_sequences(tok.texts_to_sequences([seed]), maxlen=X.shape[1])
    pred = model.predict(seq, verbose=0)
    word = tok.index_word[pred.argmax()]

    if word == "":
      break

    seed += " " + word

print(seed)
```
